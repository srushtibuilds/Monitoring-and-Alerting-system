```python
import nltk
from nltk.corpus import wordnet as wn
from nltk.tokenize import word_tokenize
from nltk.wsd import lesk

nltk.download('punkt')
nltk.download('wordnet')
nltk.download('punkt_tab') # Download punkt_tab

sentence = "I went to the bank to deposit money."
tokens = word_tokenize(sentence)
word_to_disambiguate = 'bank'

sense = lesk(tokens, word_to_disambiguate)

print(f"Chosen sense for the word '{word_to_disambiguate}':")
print(f"Synset: {sense}")
print(f"Definition: {sense.definition()}")
```

```python
import nltk
from nltk.corpus import wordnet as wn
from nltk.tokenize import word_tokenize
from nltk.wsd import lesk

nltk.download('punkt')
nltk.download('wordnet')

```

```python
def analyze_senses(sentence, target_word):
    tokens = word_tokenize(sentence)
    sense = lesk(tokens, target_word)
    return (sense, sense.definition())
#eg1
sentences = [
    ("I went to the bank to deposit money.", "bank"),
    ("He leaned against the bank of the river.", "bank"),
    ("She saw a bat flying in the dark.", "bat"),
    ("The bat hit the ball out of the park.", "bat"),
    ("He caught a cold and went to the doctor.", "cold"),
    ("The weather is cold today.", "cold")
]

results = []

for sentence, word in sentences:
    sense, definition = analyze_senses(sentence, word)
    results.append((sentence, word, sense, definition))

for sentence, word, sense, definition in results:
    print(f"Sentence: {sentence}")
    print(f"Word: {word}")
    print(f"Chosen sense: {sense}")
    print(f"Definition: {definition}")
    print("-" * 50)
```
